```python
# pip install openmeteo-requests requests-cache retry-requests numpy pandas tensorflow scikit-learn optuna

import pandas as pd
import numpy as np
from sklearn.preprocessing import RobustScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import openmeteo_requests
import requests_cache
from retry_requests import retry
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
import matplotlib.pyplot as plt
import requests
import optuna
from optuna.pruners import MedianPruner
from optuna.samplers import TPESampler
import json
import pickle           # ← added for saving scaler and feature columns
import warnings
warnings.filterwarnings('ignore')

# =============================================================================
# CONFIGURATION
# =============================================================================

ED_DATA_PATH        = "path/to/VestfoldTriageReport.csv"
INFECTION_DATA_PATH = "path/to/Infeksjonsdata.xlsx"
SAVE_DIR            = "path/to/outputs/"

HOSPITAL_LAT   = 59.2725
HOSPITAL_LON   = 10.4184
LOOKBACK       = 24
HORIZONS       = 24
EVAL_HORIZONS  = [1, 2, 6, 12, 24]

USE_OPTUNA     = True
OPTUNA_TRIALS  = 50
OPTUNA_TIMEOUT = 7200

# =============================================================================
# WEIGHTED MSE LOSS
# =============================================================================

HORIZON_WEIGHTS = tf.constant(
    np.linspace(1.0, 0.3, HORIZONS).astype(np.float32),
    dtype=tf.float32
)

def weighted_mse(y_true, y_pred):
    squared_errors = tf.square(y_true - y_pred)
    weighted       = squared_errors * HORIZON_WEIGHTS
    return tf.reduce_mean(weighted)

def weighted_mae(y_true, y_pred):
    abs_errors = tf.abs(y_true - y_pred)
    weighted   = abs_errors * HORIZON_WEIGHTS
    return tf.reduce_mean(weighted)

# =============================================================================
# DATA PIPELINE
# =============================================================================

def load_ed_data(filepath, separator=';'):
    print("\n[1/9] Loading ED data...")
    df = pd.read_csv(filepath, sep=separator)
    df.columns = ['arrival', 'departure', 'first_doctor_response', 'first_triage']
    print(f"   Loaded {len(df)} records")
    return df

def parse_datetime_columns(df):
    print("\n[2/9] Parsing datetime fields...")
    df['arrival']   = pd.to_datetime(df['arrival'],   format='%d.%m.%Y %H:%M', errors='coerce')
    df['departure'] = pd.to_datetime(df['departure'], format='%d.%m.%Y %H:%M', errors='coerce')
    df = df.dropna(subset=['arrival','departure']).sort_values('arrival')
    tz = 'Europe/Oslo'
    df['arrival']   = df['arrival'].dt.tz_localize(tz).dt.tz_convert('UTC').dt.tz_localize(None)
    df['departure'] = df['departure'].dt.tz_localize(tz).dt.tz_convert('UTC').dt.tz_localize(None)
    df['duration_hour'] = (df['departure'] - df['arrival']).dt.total_seconds() / 3600
    print(f"   Date range: {df['arrival'].min()} to {df['arrival'].max()}")
    return df

def aggregate_to_hourly(df):
    print("\n[3/9] Aggregating to hourly intervals...")
    df['arrival_hour'] = df['arrival'].dt.floor('H')
    hourly = df.groupby('arrival_hour').agg(
        arrival_count = ('arrival',       'count'),
        duration_mean = ('duration_hour', 'mean'),
        duration_std  = ('duration_hour', 'std'),
    ).reset_index()
    hourly[['duration_mean','duration_std']] = hourly[['duration_mean','duration_std']].fillna(0)
    hourly['first_doctor_response'] = df['first_doctor_response'].fillna('Unknown').values[:len(hourly)]
    hourly['first_triage']          = df['first_triage'].fillna('Unknown').values[:len(hourly)]
    print(f"   Created {len(hourly)} hourly records")
    return hourly

def fetch_weather_data(start_date, end_date):
    print(f"\n[4/9] Fetching weather data...")
    cache_session = requests_cache.CachedSession('.cache', expire_after=-1)
    retry_session = retry(cache_session, retries=5, backoff_factor=0.2)
    openmeteo     = openmeteo_requests.Client(session=retry_session)
    params = {
        "latitude": HOSPITAL_LAT, "longitude": HOSPITAL_LON,
        "start_date": start_date, "end_date": end_date, "timezone": "UTC",
        "hourly": ["temperature_2m","relative_humidity_2m","precipitation",
                   "surface_pressure","wind_speed_10m"]
    }
    response = openmeteo.weather_api(
        "https://archive-api.open-meteo.com/v1/archive", params=params)[0]
    hourly = response.Hourly()
    timestamps = pd.date_range(
        start=pd.to_datetime(hourly.Time(), unit="s"),
        end=pd.to_datetime(hourly.TimeEnd(), unit="s"),
        freq=pd.Timedelta(seconds=hourly.Interval()), inclusive="left"
    )
    weather_df = pd.DataFrame({
        "date":          timestamps,
        "temperature":   hourly.Variables(0).ValuesAsNumpy().astype(float).round(2),
        "humidity":      hourly.Variables(1).ValuesAsNumpy().astype(float).round(2),
        "precipitation": hourly.Variables(2).ValuesAsNumpy().astype(float).round(2),
        "pressure":      hourly.Variables(3).ValuesAsNumpy().astype(float).round(2),
        "wind_speed":    hourly.Variables(4).ValuesAsNumpy().astype(float).round(2),
    })
    weather_df['date'] = pd.to_datetime(weather_df['date']).dt.tz_localize(None)
    print(f"   Fetched {len(weather_df)} records")
    return weather_df

def merge_weather_data(df, weather_df):
    df['arrival_hour'] = pd.to_datetime(df['arrival_hour'])
    weather_df['date'] = pd.to_datetime(weather_df['date'])
    merged = pd.merge_asof(
        df.sort_values("arrival_hour"), weather_df.sort_values("date"),
        left_on="arrival_hour", right_on="date", direction="nearest"
    ).drop(columns=['date'], errors='ignore')
    print(f"   Merged to {len(merged)} records")
    return merged

def load_merge_infection_data(df):
    print("\n[5/9] Loading infection data...")
    monthly_df = pd.read_excel(INFECTION_DATA_PATH, header=0)
    monthly_df['Month']      = pd.to_datetime(monthly_df['Month'], format='%b-%y')
    df['year_month']         = df['arrival_hour'].dt.to_period('M')
    monthly_df['year_month'] = monthly_df['Month'].dt.to_period('M')
    df_merged = df.merge(
        monthly_df[['year_month','Total_Infected_Patient_Monthly']],
        on='year_month', how='left'
    )
    df_merged['infection_rate_hourly'] = (
        df_merged['Total_Infected_Patient_Monthly'] /
        (df_merged['arrival_hour'].dt.days_in_month * 24)
    )
    return df_merged.drop(columns=['Total_Infected_Patient_Monthly','year_month'])

def fetch_norwegian_holidays(start_year, end_year):
    all_holidays = []
    for year in range(start_year, end_year + 1):
        try:
            r = requests.get(
                f"https://date.nager.at/api/v3/PublicHolidays/{year}/NO", timeout=10)
            if r.status_code == 200:
                for h in r.json():
                    all_holidays.append({'date': pd.to_datetime(h['date'])})
        except:
            pass
    return (pd.DataFrame(all_holidays).drop_duplicates(subset=['date'])
            if all_holidays else pd.DataFrame(columns=['date']))

def add_holiday_feature(df):
    holidays_df = fetch_norwegian_holidays(
        df['arrival_hour'].dt.year.min(), df['arrival_hour'].dt.year.max())
    if holidays_df.empty:
        df['is_holiday'] = 0
    else:
        df['_date']          = pd.to_datetime(df['arrival_hour']).dt.date
        holidays_df['_date'] = holidays_df['date'].dt.date
        df['is_holiday']     = df['_date'].isin(set(holidays_df['_date'])).astype(int)
        df = df.drop(columns=['_date'])
    return df

def create_temporal_features(df):
    print("\n[6/9] Creating temporal features...")
    df['hour']        = df['arrival_hour'].dt.hour
    df['hour_sin']    = np.sin(2*np.pi*df['hour']/24)
    df['hour_cos']    = np.cos(2*np.pi*df['hour']/24)
    df['day']         = df['arrival_hour'].dt.day
    df['dayofweek']   = df['arrival_hour'].dt.dayofweek
    df['week']        = df['arrival_hour'].dt.isocalendar().week
    df['month']       = df['arrival_hour'].dt.month
    df['year']        = df['arrival_hour'].dt.year
    df['day_of_year'] = df['arrival_hour'].dt.dayofyear
    df['is_weekend']  = df['dayofweek'].isin([5,6]).astype(int)
    df['is_monday']   = (df['dayofweek']==0).astype(int)
    df['is_friday']   = (df['dayofweek']==4).astype(int)
    for col, period in [('day',31),('dayofweek',7),('week',52),('month',12)]:
        df[f'{col}_sin'] = np.sin(2*np.pi*df[col]/period)
        df[f'{col}_cos'] = np.cos(2*np.pi*df[col]/period)
    df['time_of_day'] = pd.cut(df['hour'], bins=[-1,6,12,18,24],
                               labels=['night','morning','afternoon','evening'])
    df['shift']       = pd.cut(df['hour'], bins=[-1,8,16,24],
                               labels=['night_shift','day_shift','evening_shift'])
    df['same_hour_last_week']    = df['arrival_count'].shift(168)
    df['same_hour_2weeks_ago']   = df['arrival_count'].shift(336)
    df['diff_from_last_week']    = df['arrival_count'] - df['same_hour_last_week']
    df['arrival_change_1h']      = df['arrival_count'].diff(1)
    df['arrival_change_3h']      = df['arrival_count'].diff(3)
    df['arrival_pct_change_1h']  = df['arrival_count'].pct_change(1)
    df['arrival_pct_change_24h'] = df['arrival_count'].pct_change(24)
    df['ema_12h'] = df['arrival_count'].ewm(span=12).mean()
    df['ema_24h'] = df['arrival_count'].ewm(span=24).mean()
    for col in ['first_doctor_response','first_triage','time_of_day','shift']:
        df[col] = df[col].astype('category').cat.codes
    return df

def create_lag_features(df, lags=[1,6,12,24]):
    for lag in lags:
        df[f'arrival_count_lag_{lag}'] = df['arrival_count'].shift(lag)
    return df.dropna()

# =============================================================================
# RUN PIPELINE
# =============================================================================

df = load_ed_data(ED_DATA_PATH)
df = parse_datetime_columns(df)
df = aggregate_to_hourly(df)
weather_df = fetch_weather_data("2023-01-10", "2024-10-26")
df = merge_weather_data(df, weather_df)
df = load_merge_infection_data(df)
df = add_holiday_feature(df)
df = create_temporal_features(df)
df = create_lag_features(df, lags=[1,6,12,24])
print(f"\nFinal dataset shape: {df.shape}")

# =============================================================================
# SEQUENCE PREPARATION
# =============================================================================

print("\n[7/9] Preparing sequences for direct 24-step forecasting...")

exclude_cols  = ['arrival_hour', 'arrival_count']
feature_cols  = [c for c in df.columns if c not in exclude_cols]
features      = df[feature_cols].values
target_series = df['arrival_count'].values

# ── SAVE 1: feature column names ──────────────────────────────────────────────
# Must be saved before any transformation so the order is preserved exactly
# as the model expects at inference time.
FEATURE_COLS_PATH = f"{SAVE_DIR}feature_columns.pkl"
with open(FEATURE_COLS_PATH, "wb") as f:
    pickle.dump(feature_cols, f)
print(f"\n   Saved feature columns ({len(feature_cols)} features) → {FEATURE_COLS_PATH}")

# ── Scale features with RobustScaler (inputs only — targets stay as raw counts)
scaler_X        = RobustScaler()
features_scaled = scaler_X.fit_transform(features)

# ── SAVE 2: fitted RobustScaler ───────────────────────────────────────────────
# Must be saved immediately after fit_transform on the TRAINING portion.
# NOTE: scaler_X is currently fitted on the full dataset here. If you want
# strict train-only fitting, move this block to after the split and refit
# on features[:train_end] only. The version below matches your existing pipeline.
SCALER_PATH = f"{SAVE_DIR}robust_scaler.pkl"
with open(SCALER_PATH, "wb") as f:
    pickle.dump(scaler_X, f)
print(f"   Saved RobustScaler → {SCALER_PATH}")

print(f"\n   Target stats (raw, unscaled):")
print(f"   Mean: {target_series.mean():.2f}  Std: {target_series.std():.2f}"
      f"  Min: {target_series.min():.0f}  Max: {target_series.max():.0f}")
print(f"   → Training on raw counts (no target scaling, zeros kept intact)")

# Build X (lookback window) and Y (next 24 raw counts)
X_all, Y_all = [], []
n = len(features_scaled)
for i in range(LOOKBACK, n - HORIZONS + 1):
    X_all.append(features_scaled[i - LOOKBACK : i])
    Y_all.append(target_series[i : i + HORIZONS].astype(np.float32))

X_all = np.array(X_all, dtype=np.float32)
Y_all = np.array(Y_all, dtype=np.float32)

print(f"\n   X shape: {X_all.shape}")
print(f"   Y shape: {Y_all.shape}")

# 80 / 10 / 10 chronological split
n_samples = len(X_all)
train_end = int(n_samples * 0.80)
val_end   = int(n_samples * 0.90)

X_train, Y_train = X_all[:train_end],        Y_all[:train_end]
X_val,   Y_val   = X_all[train_end:val_end], Y_all[train_end:val_end]
X_test,  Y_test  = X_all[val_end:],          Y_all[val_end:]

print(f"\n   Train:      {len(X_train)} ({len(X_train)/n_samples*100:.1f}%)")
print(f"   Validation: {len(X_val)}   ({len(X_val)/n_samples*100:.1f}%)")
print(f"   Test:       {len(X_test)}  ({len(X_test)/n_samples*100:.1f}%)")

# =============================================================================
# MODEL BUILDER
# =============================================================================

def build_direct_lstm(params, input_shape, n_outputs=24):
    model = keras.Sequential(name='Direct_LSTM_Optuna_24h')

    n_layers = params['n_layers']
    for i in range(n_layers):
        return_sequences = (i < n_layers - 1)
        lstm_kwargs = dict(
            units=params[f'hidden_size_{i}'],
            return_sequences=return_sequences,
            kernel_regularizer=keras.regularizers.l2(params['l2_reg']),
        )
        if i == 0:
            lstm_kwargs['input_shape'] = input_shape
        model.add(layers.LSTM(**lstm_kwargs))
        model.add(layers.Dropout(params['dropout_rate']))
        if params['batch_norm']:
            model.add(layers.BatchNormalization())

    n_dense = params['dense_layers']
    for i in range(n_dense):
        model.add(layers.Dense(
            params[f'dense_units_{i}'], activation='relu',
            kernel_regularizer=keras.regularizers.l2(params['l2_reg'])
        ))
        model.add(layers.Dropout(params['dropout_rate'] * 0.3))

    model.add(layers.Dense(n_outputs, activation='linear', name='output_24h'))

    opt = (keras.optimizers.Adam(params['learning_rate'])
           if params['optimizer'] == 'adam'
           else keras.optimizers.RMSprop(params['learning_rate']))

    model.compile(
        optimizer=opt,
        loss=weighted_mse,
        metrics=[weighted_mae]
    )
    return model

# =============================================================================
# OPTUNA SEARCH
# =============================================================================

ALL_HIDDEN = [32, 64, 128, 256]
ALL_DENSE  = [16, 32, 64, 128]

if USE_OPTUNA:
    print("\n" + "="*70)
    print("OPTUNA HYPERPARAMETER SEARCH — Direct 24-output LSTM")
    print("="*70)

    def optuna_objective(trial):
        keras.backend.clear_session()

        n_layers = trial.suggest_int('n_layers', 1, 3)
        hidden_sizes = []
        for i in range(n_layers):
            raw = trial.suggest_categorical(f'hidden_size_{i}', ALL_HIDDEN)
            if i > 0:
                raw = min(raw, hidden_sizes[i-1])
            hidden_sizes.append(raw)

        n_dense = trial.suggest_int('dense_layers', 1, 2)
        dense_units = [
            trial.suggest_categorical(f'dense_units_{i}', ALL_DENSE)
            for i in range(n_dense)
        ]

        params = {
            'n_layers':     n_layers,
            **{f'hidden_size_{i}': hidden_sizes[i] for i in range(n_layers)},
            'dense_layers': n_dense,
            **{f'dense_units_{i}': dense_units[i] for i in range(n_dense)},
            'dropout_rate':  trial.suggest_float('dropout_rate', 0.1, 0.4, step=0.05),
            'learning_rate': trial.suggest_float('learning_rate', 1e-4, 5e-3, log=True),
            'l2_reg':        trial.suggest_categorical('l2_reg',
                                 [0.001, 0.005, 0.01, 0.02]),
            'batch_norm':    trial.suggest_categorical('batch_norm', [True, False]),
            'optimizer':     trial.suggest_categorical('optimizer', ['adam', 'rmsprop']),
            'batch_size':    trial.suggest_categorical('batch_size', [16, 32, 64]),
        }

        input_shape = (X_train.shape[1], X_train.shape[2])
        model = build_direct_lstm(params, input_shape, n_outputs=HORIZONS)

        early_stop = EarlyStopping(
            monitor='val_loss', patience=10,
            restore_best_weights=True, mode='min'
        )
        try:
            history = model.fit(
                X_train, Y_train,
                validation_data=(X_val, Y_val),
                epochs=60,
                batch_size=params['batch_size'],
                callbacks=[early_stop],
                shuffle=False,
                verbose=0
            )
            return min(history.history['val_loss'])
        except Exception as e:
            print(f"  Trial failed: {e}")
            raise optuna.exceptions.TrialPruned()

    study = optuna.create_study(
        direction='minimize',
        sampler=TPESampler(seed=42),
        pruner=MedianPruner(n_startup_trials=10, n_warmup_steps=5)
    )

    try:
        study.optimize(
            optuna_objective,
            n_trials=OPTUNA_TRIALS,
            timeout=OPTUNA_TIMEOUT,
            show_progress_bar=True,
            catch=(Exception,)
        )
        if study.best_trial:
            print(f"\n  Best trial: #{study.best_trial.number}")
            print(f"  Best val loss (weighted MSE): {study.best_trial.value:.6f}")
            best_params = study.best_params
        else:
            raise ValueError("No successful trials")
    except Exception as e:
        print(f"\n  Search failed ({e}) — using sensible defaults")
        best_params = {
            'n_layers': 2, 'hidden_size_0': 128, 'hidden_size_1': 64,
            'dense_layers': 2, 'dense_units_0': 64, 'dense_units_1': 32,
            'dropout_rate': 0.30, 'learning_rate': 0.0004,
            'l2_reg': 0.01, 'batch_norm': True,
            'optimizer': 'rmsprop', 'batch_size': 32,
        }
else:
    print("\n  Skipping Optuna — using sensible defaults")
    best_params = {
        'n_layers': 2, 'hidden_size_0': 128, 'hidden_size_1': 64,
        'dense_layers': 2, 'dense_units_0': 64, 'dense_units_1': 32,
        'dropout_rate': 0.30, 'learning_rate': 0.0004,
        'l2_reg': 0.01, 'batch_norm': True,
        'optimizer': 'rmsprop', 'batch_size': 32,
    }

PARAMS_PATH = f"{SAVE_DIR}best_params_direct_lstm.json"
with open(PARAMS_PATH, 'w') as f:
    json.dump(best_params, f, indent=4)
print(f"\n  Best hyperparameters saved → {PARAMS_PATH}")
for k, v in best_params.items():
    print(f"    {k}: {v}")

# =============================================================================
# FINAL TRAINING
# =============================================================================

print("\n[8/9] Final training with best parameters...")
print("="*70)

input_shape = (X_train.shape[1], X_train.shape[2])
model = build_direct_lstm(best_params, input_shape, n_outputs=HORIZONS)
model.summary()

callbacks = [
    EarlyStopping(monitor='val_loss', patience=20,
                  restore_best_weights=True, verbose=1, mode='min'),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=10,
                      min_lr=1e-7, verbose=1, mode='min'),
    ModelCheckpoint('best_direct_lstm_24h.h5', monitor='val_loss',
                    save_best_only=True, verbose=1, mode='min'),
]

history = model.fit(
    X_train, Y_train,
    validation_data=(X_val, Y_val),
    epochs=150,
    batch_size=best_params['batch_size'],
    callbacks=callbacks,
    shuffle=False,
    verbose=1
)

# ── SAVE 3: final trained model in .keras format ──────────────────────────────
# .keras is the recommended modern format (replaces .h5).
# Saves the full model: architecture + weights + optimizer state + custom
# loss/metric functions. Required for Hugging Face deployment.
KERAS_PATH = f"{SAVE_DIR}direct_lstm_optuna.keras"
model.save(KERAS_PATH)
print(f"\n   Saved Keras model → {KERAS_PATH}")

# Also keep the .h5 checkpoint already saved by ModelCheckpoint
print(f"   Also saved (ModelCheckpoint) → best_direct_lstm_24h.h5")

# =============================================================================
# SUMMARY OF SAVED ARTEFACTS
# =============================================================================

import os
artefacts = {
    "Feature columns": FEATURE_COLS_PATH,
    "RobustScaler":    SCALER_PATH,
    "Keras model":     KERAS_PATH,
    "Best params":     PARAMS_PATH,
}
print("\n" + "="*70)
print("SAVED ARTEFACTS")
print("="*70)
for name, path in artefacts.items():
    size_mb = os.path.getsize(path) / 1e6 if os.path.exists(path) else 0
    print(f"  {name:<20} {path}")
    print(f"  {'':20} Size: {size_mb:.2f} MB")
print()
print("  Upload all four files to your Hugging Face Space.")
print("  The app.py loads them with:")
print("    model      = tf.keras.models.load_model('direct_lstm_optuna.keras')")
print("    scaler     = pickle.load(open('robust_scaler.pkl', 'rb'))")
print("    feat_cols  = pickle.load(open('feature_columns.pkl', 'rb'))")

# =============================================================================
# EVALUATION
# =============================================================================

print("\n[9/9] Evaluating on test set...")

Y_pred = model.predict(X_test, verbose=0)
Y_pred = np.clip(Y_pred, 0, None)
Y_true = Y_test

def metrics_at_horizon(y_true_all, y_pred_all, h_idx):
    yt = y_true_all[:, h_idx]
    yp = y_pred_all[:, h_idx]
    rmse = np.sqrt(mean_squared_error(yt, yp))
    mae  = mean_absolute_error(yt, yp)
    r2   = r2_score(yt, yp)
    nonzero = yt > 0
    mape = np.mean(np.abs((yt[nonzero] - yp[nonzero])
                           / yt[nonzero])) * 100
    return rmse, mae, r2, mape

print(f"\n{'='*70}")
print(f"  DIRECT LSTM+Optuna — per-horizon test results")
print(f"{'='*70}")
print(f"  {'Horizon':>8}  {'RMSE':>7}  {'MAE':>7}  {'R²':>8}  {'MAPE':>7}")
print(f"  {'-'*55}")

summary_rows = []
for h in EVAL_HORIZONS:
    rmse, mae, r2, mape = metrics_at_horizon(Y_true, Y_pred, h_idx=h-1)
    print(f"  {h:>6}h  {rmse:>7.3f}  {mae:>7.3f}  {r2:>8.4f}  {mape:>6.2f}%")
    summary_rows.append({'Horizon': f'{h}h', 'RMSE': round(rmse,3),
                         'MAE': round(mae,3), 'R2': round(r2,4),
                         'MAPE': round(mape,2)})

summary_df = pd.DataFrame(summary_rows)
summary_df.to_csv(f'{SAVE_DIR}direct_lstm_optuna_results.csv', index=False)
print(f"\n  Saved: direct_lstm_optuna_results.csv")

# =============================================================================
# VISUALISATION
# =============================================================================

for h in EVAL_HORIZONS:
    h_idx   = h - 1
    preds   = Y_pred[:48, h_idx]
    actuals = Y_true[:48, h_idx]
    x       = np.arange(len(preds))
    rmse, mae, r2, mape = metrics_at_horizon(Y_true, Y_pred, h_idx)

    fig, ax = plt.subplots(figsize=(14, 6))
    ax.bar(x, actuals, alpha=0.5, label='Actual Patient Count',
           color='#2E86AB', edgecolor='black')
    ax.plot(x, preds, 'o-', label='Predicted Patient Count',
            color='#E8451E', linewidth=2, markersize=6)
    ax.set_xlabel('Hour', fontsize=11)
    ax.set_ylabel('Patient Arrivals', fontsize=11)
    ax.set_title(
        f'Direct LSTM+Optuna — {h}-Hour Ahead Forecast\n'
        f'RMSE={rmse:.3f}, MAE={mae:.3f}, R²={r2:.4f}',
        fontweight='bold'
    )
    ax.legend(fontsize=11)
    ax.grid(True, alpha=0.3, linestyle='--')
    ax.set_xticks(x)
    plt.tight_layout()
    fname = f'{SAVE_DIR}DirectLSTMOptuna_forecast_{h}h.png'
    fig.savefig(fname, dpi=300, bbox_inches='tight')
    print(f"  Saved: {fname}")
    plt.show()

# Training history
fig2, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(history.history['loss'],     label='Train (weighted MSE)')
axes[0].plot(history.history['val_loss'], label='Val (weighted MSE)')
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Weighted MSE')
axes[0].set_title('Training vs Validation Loss')
axes[0].legend(); axes[0].grid(True, alpha=0.3)

axes[1].plot(history.history['weighted_mae'],     label='Train (weighted MAE)')
axes[1].plot(history.history['val_weighted_mae'], label='Val (weighted MAE)')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Weighted MAE')
axes[1].set_title('Training vs Validation MAE')
axes[1].legend(); axes[1].grid(True, alpha=0.3)

plt.tight_layout()
fig2.savefig(f'{SAVE_DIR}DirectLSTMOptuna_training_history.png',
             dpi=300, bbox_inches='tight')
print(f"  Saved: DirectLSTMOptuna_training_history.png")
plt.show()

print("\n" + "="*70)
print("DONE — Direct 24-output LSTM+Optuna complete.")
print("="*70)
```
